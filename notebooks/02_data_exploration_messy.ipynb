```python
import pandas as pd
```

```python
messy_Data_file_path=  "/Users/bibashnepali/Desktop/AI_Data_Intelligence/ai_data_intelligence/data/messy_ecommerce_sales_data.csv"
messy_data = pd.read_csv(messy_Data_file_path)
```

```python
df = pd.DataFrame(messy_data)
```

```python
df.info()
```

```python
#checking for missing values
df.isnull().sum()
if (df.isnull().sum().sum() == 0):
    print("No missing values in the dataset.")
else:
    print("There are missing values in the dataset.")
```

```python
#checking for duplicates
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")
```

```python
#removing duplicates if duplicate_count > 0:
if duplicate_count > 0:
    df = df.drop_duplicates()
    print(f"Duplicates removed. New number of rows: {len(df)}")
```

```python
#removing the null values
df = df.dropna()
print(f"Null values removed. New number of rows: {len(df)}")
```
